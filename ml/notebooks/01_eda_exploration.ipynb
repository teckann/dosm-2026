# DOSM Datathon 2026: Exploratory Data Analysis (EDA)
This notebook explores the socio-economic indicators across Malaysian states, analyzing income distributions, poverty incidence, and digital adoption trends.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

sns.set_theme(style="whitegrid")
%matplotlib inline
```

```python
# Load raw dataset
DATA_PATH = Path("../../data/raw/sample_dataset.csv")
df = pd.read_csv(DATA_PATH)
print(f"Dataset Shape: {df.shape}")
df.head(10)
```

```python
# Summary statistics
df.describe().round(2)
```

```python
# Correlation analysis
numeric_cols = df.select_dtypes(include=[np.number]).columns
plt.figure(figsize=(10, 6))
sns.heatmap(df[numeric_cols].corr(), annot=True, cmap="coolwarm", fmt=".2f", cbar=True)
plt.title("Correlation Matrix of Socio-Economic Indicators (DOSM 2026)")
plt.show()
```

```python
# Median Income vs Poverty Rate by State (Latest Year)
latest_df = df[df["year"] == df["year"].max()]

plt.figure(figsize=(12, 6))
sns.scatterplot(
    data=latest_df,
    x="median_income",
    y="poverty_rate",
    size="population_millions",
    hue="state",
    sizes=(100, 500),
    legend="full"
)
plt.title("Median Income vs Poverty Rate across Malaysian States")
plt.xlabel("Median Household Income (RM)")
plt.ylabel("Poverty Rate (%)")
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

### Key Observations
1. Strong negative non-linear relationship between Median Household Income and Poverty Rate.
2. Higher digital adoption index correlates with lower unemployment and higher socio-economic resilience.
3. Ready for feature engineering and model training in `ml/src/train.py`.
